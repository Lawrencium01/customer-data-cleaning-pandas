# Converting Data Types

```python
num_int = 7

type(num_int)
```

```python
num_str = '7'

type(num_str)
```

```python
num_str_conv = int(num_str)

type(num_str_conv)
num_str_conv
```

```python
num_sum = num_int + num_str_conv

print(num_sum)
```

```python
type(num_sum)
```
