# If - Elif - Else Statements

```python
if (25 > 10):
    print('It worked!')
```

```python
if (25 < 10) or (1 < 3):
    print('It worked!')
elif 25 < 20:
    print('Elif worked!')
elif 25 < 21:
    print('Elif 2 worked!')
elif 25 < 40:
    print('Elif 3 worked!')
elif 25 < 50:
    print('Elif 4 worked!')
else:
    print('It did not work...')
```

```python
print('It worked!') if 10 > 30 else print('It did not work...')
```

```python
if (25 < 10) or (1 < 3):
    print('It worked!')
    if 10 > 11:
        print('This nested if statement worked!')
    else:
        print('This nested if statement did not work...')
elif 25 < 20:
    print('Elif worked!')
elif 25 < 21:
    print('Elif 2 worked!')
elif 25 < 40:
    print('Elif 3 worked!')
elif 25 < 50:
    print('Elif 4 worked!')
else:
    print('It did not work...')
```
