# Comparison Operator

```python
10 == 50
```

```python
10 != 50
```

```python
'Vanilla' == 'Vanilla'
```

```python
x = 'Vanilla'
y = 'Chocolate'

x != y
```

```python
50 >= 10
```

# Logical Operators

```python
(70 > 50) and (50 > 10)
```

```python
(10 > 50) or (50 > 10)
```

```python
('Vanilla' > 'Chocolate') and (50 > 10)
```

```python
not(50 > 10)
```

# Membership Operator

```python
ice_cream = 'I love Chocolate Ice cream'
'love' in ice_cream
```

```python
scoops = [1,2,3,4,5]
6 not in scoops
```

```python
scoops = [1,2,3,4,5]
wanted_scoop = 8
wanted_scoop in scoops
```
