```python
# sets


daily_pints = {1,2,3}
```

```python
type(daily_pints)
```

```python
print(daily_pints)
```

```python
daily_pints_log = {1,2,31,5,6,3,2,1,3}
```

```python
print(daily_pints_log)
```

```python
wifes_daily_pints_log = {1,3,2,4,5,6,35,7,8,3,2,1}
```

```python
print(daily_pints_log | wifes_daily_pints_log)
```

```python
print(daily_pints_log & wifes_daily_pints_log)
```

```python
print(wifes_daily_pints_log - daily_pints_log)
```

```python
print(wifes_daily_pints_log ^ daily_pints_log)
```
